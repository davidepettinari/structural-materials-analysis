```python
import matplotlib.pyplot as plt

# Dati materiali: (nome, anno, temperatura massima operativa in °C)
materials = [
    ("Steel 660", 1940, 970), #
    ("SS316", 1960, 1140), #
    ("Inconel 718", 1960, 970), #
    ("Inconel 625", 1965, 1250), #
    ("XM-19", 1975, 700), #
    ("SiC/SiC", 1980, 1870), #
    ("G91 T1", 1980, 870), #
    ("G91 T2", 1985, 900), #
    ("F82H", 1990, 820), #
    ("V-4Cr-4Ti", 1994, 920),
    ("EUROFER 97", 1996, 820), #
    ("CLAM", 2000, 820), #
    ("RUSFER", 2006, 1150), #
    ("HiperFer", 2015, 920), #
    ("SS316L", 1970, 840),
    ("HT9", 1965, 900),
    ("ODS-EUROFER", 2002, 920), 
]

# Temperature dei materiali del breeder e refrigeranti
breeder_temperatures = {
    "DEMO: Water cooled Lead Ceramic Breeder (~570 K)": 570,
    "STEP: Encapsulated breeding blanket (~670 K)": 670,
    "ITER: Helium-cooled lithium lead (~750 K)": 750,
    "ARC: FLiBe (~850 K)": 850,
}

# Creazione del grafico
plt.figure(figsize=(12, 7))
for name, year, temp in materials:
    plt.plot(year, temp, 'o', markersize=8)
    plt.text(year, temp + 10, name, ha='center', va='bottom', fontsize=10, rotation=45)


# Aggiunta delle linee e aree colorate per i materiali del breeder
colors = ['lightblue','green', 'gold', 'red']
for (label, temp), color in zip(breeder_temperatures.items(), colors):
    plt.axhline(y=temp, color=color, linestyle='--', label=label)
    plt.ylim(500, 2020)
    plt.xlim(1935, 2020)
    plt.xticks(fontsize=12)
    plt.yticks(fontsize=12)

# Etichette e legenda
plt.xlabel("Year of Introduction", fontsize=15)
plt.ylabel("Maximum Operating Temperature [K]", fontsize=15)
plt.grid(True)
# plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.legend(loc='upper left', fontsize=12)

plt.tight_layout()
plt.savefig('fig.pdf', format='pdf', bbox_inches='tight')

plt.show()
```
