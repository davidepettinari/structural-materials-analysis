```python
import matplotlib.pyplot as plt

# Dati materiali: (nome, anno, temperatura massima operativa in °C)
materials = [
    ("SS316", 1960, 600),
    ("Inconel 718", 1965, 700),
    ("F82H", 1995, 550),
    ("Eurofer", 2000, 550),
    ("CLAM", 2002, 550),
    ("Rusfer", 2006, 550),
    ("Hyperfer", 2015, 600),
    ("SiC/SiC", 2020, 1000),
    ("XM-19", 1960, 600),
    ("Steel 660", 1950, 650),
    ("G91 T1", 1985, 600),
    ("G91 T2", 1990, 600),
    ("Ti alloy", 1970, 600),
    ("Cr alloy", 1980, 700)
]

# Temperature dei materiali del breeder e refrigeranti
breeder_temperatures = {
    "ITER: Helium-cooled lithium lead (~550°C)": 750,
    "DEMO: Water cooled Lead Ceramic Breeder (~570 K)": 570,
    "ARC: FLiBe (~850 K)": 850,
    "STEP: Encapsulated breeding blanket (~670 K)": 670
}

# Creazione del grafico
plt.figure(figsize=(12, 7))
for name, year, temp in materials:
    plt.plot(year, temp, 'o', label=name)

# Aggiunta delle linee e aree colorate per i materiali del breeder
colors = ['lightcoral', 'gold', 'lightblue']
for (label, temp), color in zip(breeder_temperatures.items(), colors):
    plt.axhline(y=temp, color=color, linestyle='--', label=label)
    plt.fill_between([1950, 2025], temp, 1100, color=color, alpha=0.4)

# Etichette e legenda
plt.xlabel("Year of Introduction")
plt.ylabel("Maximum Operating Temperature [K]")
plt.title("Maximum Operating Temperature of Structural Materials vs. Year of Introduction")
plt.grid(True)
# plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.legend(loc='upper left')

plt.tight_layout()

plt.show()
```
